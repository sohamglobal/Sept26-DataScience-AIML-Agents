```python
import pandas
from sklearn.cluster import KMeans
```

```python
url="https://raw.githubusercontent.com/sohamglobal/datasets/refs/heads/main/testcricketers.csv"
df=pandas.read_csv(url)
```

```python
print(df.head(3))
print(df.tail(3))
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
X=df[['TestRuns','TestWickets']]
```

```python
km=KMeans(n_clusters=3,random_state=42)
```

```python
df['cluster']=km.fit_predict(X)
```

```python
print(df)
```

```python
print(df.groupby('cluster')[['TestRuns','TestWickets']].mean())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['TestRuns'],df['TestWickets'],c=df['cluster'])
plt.xlabel('Runs')
plt.ylabel('Wickets')
plt.title("Player clusters")
plt.show()
```
