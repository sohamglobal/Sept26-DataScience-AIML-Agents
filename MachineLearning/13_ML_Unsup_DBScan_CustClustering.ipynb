```python
import pandas
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
data={
    'annual_spending':[
        10000,12000,11000,13000,9000,
        15000,80000,82000,78000,85000,
        81000,79000,50000,150000
    ],
    'visits_per_month':[
        1,2,1,3,2,2,
        8,9,8,10,9,8,4,1
    ]
}
```

```python
df=pandas.DataFrame(data)
print(df.head(3))
```

```python
X=StandardScaler().fit_transform(df)
```

```python
model=DBSCAN(eps=0.8,min_samples=3)
```

```python
df['cluster']=model.fit_predict(X)
```
