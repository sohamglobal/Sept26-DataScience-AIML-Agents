```python
import pandas
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
```

```python
df=pandas.read_csv("simplefraud.csv")
print(df.head(3))
```

```python
X=df[['Amount','TransactionsPerDay']]
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
db=DBSCAN(eps=0.8,min_samples=3)
```

```python
df['cluster']=db.fit_predict(X_scaled)
```

```python
print(df)
```

```python
print(df[df['cluster']==-1])
```
