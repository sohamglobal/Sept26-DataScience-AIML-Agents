```python
import random
```

```python
# Actions that the dog can possible take
actions=['sit','stand']
```

```python
# allot memory for rewards for each action
reward_table={
    "sit":0,
    "stand":0
}
```

```python
# I instruct the dog to "SIT"
reward=0
for episode in range(1,21):
  # Brownie chooses an action randomly
  action=random.choice(actions)

  # Give reward to Brownie for correct action
  if action=='sit':
    reward=1
  else:
    reward=0

  reward_table[action]+=reward
  print(f"Episode {episode} -> Action : {action} , Reward :{reward} ")

print("Learning result : ",reward_table)

```

```python
best_action=max(reward_table,key=reward_table.get)
print("Brownie has learned to ",best_action)
```
