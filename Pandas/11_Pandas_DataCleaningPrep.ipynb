```python
import pandas
```

```python
df=pandas.read_csv("messy_films.csv")
print(df)
```

**Understanding Data**

```python
print(df.head())
```

```python
print(df.tail())
```

```python
print(df.shape)
```

```python
print(df.info())
```

**Find Missing Values**

```python
print(df.isnull().sum())
```

**Remove unwanted spaces**

```python
df['Title']=df['Title'].str.strip()
```

```python
df['Genre']=df['Genre'].str.strip()
```

```python
print(df)
```

**Standardize text**

```python
df['Genre']=df['Genre'].str.title()
```

```python
df['Title']=df['Title'].str.title()
```

```python
print(df)
```

**Remove duplicate rows**

```python
print(df.duplicated())
```

```python
df=df.drop_duplicates()
```

```python
print(df)
```

**Replace missing data**

```python
df['Genre']=df['Genre'].fillna('Unknown')
```

```python
print(df)
```

**Remove row with null year**

```python
df=df.dropna(subset=['Year'])
```

```python
print(df)
```

```python
df['Rating']=df['Rating'].fillna(df['Rating'].mean())
```

```python
print(df)
```

**Data types**

```python
print(df.dtypes)
```

```python
# df['Year']=pandas.to_numeric(df['Year']) if the year is string
df['Year']=df['Year'].astype(int)
```

```python
print(df)
```

**Describe**

```python
print(df.describe())
```

```python
print(df['Rating'].describe())
```

**Round the rating to decimal place 1**

```python
df['Rating']=df['Rating'].round(1)
```

```python
print(df)
```

**Change Column names**

```python
df.rename(columns={
    'Budget_Crore':'Budget',
    'Collection_Crore':'Collection'
},inplace=True)
```

```python
print(df)
```

**Add calculated column**

```python
df['Profit']=(
    df['Collection']-df['Budget']
)
```

```python
print(df)
```

```python
df['Status']=df['Profit'].apply(
    lambda x: 'Hit' if x>700 else 'Average'
)
```

```python
print(df)
```

**Sort**

```python
print(df.sort_values('Rating',ascending=False))
```
