**Total balance of each account type**

```python
result=df.groupby("acctype")["balance"].sum()
print(result)
```

```python
result=df.groupby("acctype")["accnm"].count()
print(result)
```

```python
result=df.groupby("acctype")["balance"].mean()
print(result)
```

```python
result=df.groupby("acctype")["balance"].max()
print(result)
```

```python
result=df.groupby("acctype")["balance"].agg(
    ["count","sum","mean"]
)
print(result)
```
