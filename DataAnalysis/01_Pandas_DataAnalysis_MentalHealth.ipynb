**Standardize text**

```python
df["gender"] = df["gender"].astype("string").str.strip()
```

```python
df["state"] = df["state"].replace({
    "Maharastra": "Maharashtra",
    "maharashtra": "Maharashtra",
    "UP": "Uttar Pradesh"
})
```

**DATA PREPARATION**

**Convert numeric columns**

```python
df["age"] = pandas.to_numeric(df["age"], errors="coerce")
```

```python
numeric_columns = [
    "age",
    "work_hours_week",
    "study_hours_week",
    "sleep_hours",
    "exercise_days_week",
    "screen_time_hours_day",
    "social_media_hours_day",
    "job_satisfaction",
    "stress_score",
    "anxiety_score",
    "wellbeing_score",
    "life_satisfaction"
]

for col in numeric_columns:
    df[col] = pandas.to_numeric(df[col], errors="coerce")
```

**Handling invalid data**

**Convert invalid values to NaN**

```python
df.loc[
    ~df["age"].between(18,30),
    "age"
] = numpy.nan
```

```python
df.loc[
    ~df["sleep_hours"].between(0,14),
    "sleep_hours"
] = numpy.nan
```

```python
df.loc[
    ~df["exercise_days_week"].between(0,7),
    "exercise_days_week"
] = numpy.nan
```

**Missing value treatment**

**Numerical missing values**

```python
print(df["sleep_hours"].isnull().sum())
```

```python
print(df["sleep_hours"].median())
```

```python
df["sleep_hours"] = df["sleep_hours"].fillna(
    df["sleep_hours"].median()
)
```

**Categorical missing values**

```python
print(df["income_level"].mode())
```

```python
df["income_level"] = df["income_level"].fillna(
    df["income_level"].mode()[0]
)

# Mode = the value that appears most often.
```

**FEATURE ENGINEERING**

create age groups

```python
df["age_group"] = pandas.cut(
    df["age"],
    bins=[17,20,23,26,30],
    labels=["18-20","21-23","24-26","27-30"]
)
```

```python
df["sleep_category"] = pandas.cut(
    df["sleep_hours"],
    bins=[0,5.99,7.99,14],
    labels=[
        "Low (<6h)",
        "Moderate (6-8h)",
        "High (>8h)"
    ]
)
```

```python
df["screen_time_category"] = pandas.cut(
    df["screen_time_hours_day"],
    bins=[-0.01,4,8,24],
    labels=[
        "Low (≤4h)",
        "Moderate (4-8h)",
        "High (>8h)"
    ]
)
```

```python
df["exercise_category"] = pandas.cut(
    df["exercise_days_week"],
    bins=[-0.1,1,3,7],
    labels=[
        "Low (0-1)",
        "Moderate (2-3)",
        "High (4-7)"
    ]
)
```
