```python
import pandas
from matplotlib import pyplot
```

```python
data={
    "month":["Jan26","Feb26","Mar26","Apr26","may26","Jun26"],
    "collection":[24,17,35,19,22,20]
}
df=pandas.DataFrame(data)
print(df)
```

LINE chart - Growth & Decline

```python
pyplot.plot(df['month'],df['collection'],marker='o')
pyplot.xlabel("Month")
pyplot.ylabel("Collection in Thousands")
pyplot.title("Monthly collection")
pyplot.show()
```

BAR Chart - Compare values

```python
product=['laptop','mobile','tablet','smartwatch']
sale=[50,124,26,72]

pyplot.bar(product,sale)
pyplot.xlabel("Product")
pyplot.ylabel("Sale")
pyplot.title("Monthly Sale Report")
pyplot.show()
```

**PIE chart - percentage share distribution**

```python
product=['laptop','mobile','tablet','smartwatch']
sale=[50,124,26,72]

pyplot.pie(sale,labels=product,autopct="%1.1f%%")
pyplot.title("Sales share")
pyplot.show()
```

SCATTER chart - correlation, density of population

```python
area_k=[1,2,3,4,5]
price_sqft=[40,50,60,70,80]
pyplot.scatter(area_k,price_sqft)
pyplot.title('land price')
pyplot.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
num_records=150
data = {
    'Branch': np.random.choice(['North', 'South', 'East', 'West'], size=num_records),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Groceries'], size=num_records),
    'Advertising_Spend': np.round(np.random.uniform(500, 5000, size=num_records), 2),
    'Satisfaction_Score': np.random.choice([1, 2, 3, 4, 5], size=num_records, p=[0.05, 0.1, 0.25, 0.4, 0.2]),
}
df=pandas.DataFrame(data)
df['Sales'] = np.round(df['Advertising_Spend'] * 1.8 + np.random.normal(1000, 500, size=num_records), 2)
print(df.head(3))
```

```python
print(df['Branch'].value_counts())
```

```python
ser=df['Branch'].value_counts()
pyplot.pie(ser.values,labels=ser.index)
pyplot.show()
```

```python
# Total sales by Store Branch
branch_sales = df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 4))
branch_sales.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Total Sales Revenue by Branch')
plt.xlabel('Branch Location')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

# Scatter plot: Ad Spend vs. Sales
ax.scatter(df['Advertising_Spend'], df['Sales'], color='darkgreen', alpha=0.6)

# Custom annotations and styling
ax.set_title('Impact of Advertising Spend on Sales', fontsize=14, fontweight='bold')
ax.set_xlabel('Advertising Spend ($)', fontsize=12)
ax.set_ylabel('Sales ($)', fontsize=12)

# Reference line for average sales
avg_sales = df['Sales'].mean()
ax.axhline(avg_sales, color='red', linestyle='--', label=f'Avg Sales (${avg_sales:,.2f})')

ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Boxplot: Sales distributions per Category across Branches
plt.figure(figsize=(9, 5))
sns.set_theme(style="whitegrid")

sns.boxplot(data=df, x='Category', y='Sales', hue='Branch', palette='Set2')
plt.title('Sales Distribution by Product Category & Branch')
plt.xlabel('Product Category')
plt.ylabel('Sales ($)')
plt.legend(title='Branch Location')
plt.tight_layout()
plt.show()
```

```python
# Select numerical variables for correlation analysis
numeric_cols = ['Advertising_Spend', 'Satisfaction_Score', 'Sales']
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix (Synthetic Store Metrics)')
plt.tight_layout()
plt.show()
```
